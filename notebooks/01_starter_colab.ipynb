## Step 3: Sample Data Analysis Pipeline

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
num_samples = 100
num_genes = 50

genes = [f"Gene_{i+1:02d}" for i in range(num_genes)]
cell_types = ["Neuron", "Astrocyte", "Microglia", "Oligodendrocyte"]
samples = [f"Cell_{i+1:03d}" for i in range(num_samples)]
cell_type_assignments = np.random.choice(cell_types, size=num_samples)

expression_matrix = np.random.negative_binomial(n=5, p=0.3, size=(num_samples, num_genes))
df = pd.DataFrame(expression_matrix, index=samples, columns=genes)
df["CellType"] = cell_type_assignments

print("Expression Matrix Shape:", df.shape)
df.head()
```

## Step 4: Visualize & Save Results

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="CellType", y="Gene_01", palette="viridis")
plt.title("Male CNS Cell Type Expression Profile: Gene_01", fontsize=14)
plt.ylabel("Expression Level")
plt.xlabel("Cell Type")

output_fig_path = paths["results"] / "gene_01_celltype_expression.png"
plt.savefig(output_fig_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"✓ Visualization saved to {output_fig_path}")
```
